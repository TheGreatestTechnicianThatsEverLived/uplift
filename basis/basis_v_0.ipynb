```python
import pandas as pd
```

```python
df_clients = pd.read_csv('../data/clients.csv')
df_train = pd.read_csv('../data/uplift_train.csv')
df_test = pd.read_csv('../data/uplift_test.csv')
```

# Cleaning

```python
df_clients = df_clients[df_clients['age'].between(10, 90)]
```

# Join

```python
df_train['test'] = 0
df_test['test'] = 1
```

```python
df_t = pd.concat([df_train, df_test]).fillna(0)
```

```python
df = df_clients.merge(df_t, how='left')
```

```python
df.to_csv('data_saved/full_basis.csv')
```

```python
df[df['test'] == 0].to_csv('data_saved/train_basis.csv')
df[df['test'] == 1].to_csv('data_saved/test_basis.csv')
```
